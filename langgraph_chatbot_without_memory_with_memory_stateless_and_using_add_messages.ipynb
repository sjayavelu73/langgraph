```python
# This chatbot is naive . It does not have memory
#!pip install langgraph
from typing import TypedDict,Sequence,Annotated
from langgraph.graph import StateGraph,END,START
from google.colab import userdata
import os
os.environ['OPENAI_API_KEY']=userdata.get('ai_agents_openai')
#!pip install langchain_openai
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0)
from langchain_core.messages import HumanMessage


class Agentbot(TypedDict):
  messages: list[HumanMessage]

def model_invoke(state: Agentbot)-> Agentbot:
  response=llm.invoke(state['messages'])
  print(response.content)
  return(state)

graph=StateGraph(Agentbot)
graph.add_node("action",model_invoke)
graph.add_edge(START,"action")
graph.add_edge("action",END)

compiled_graph=graph.compile()

print("Starting chatbot")

user_question=input("How can i help you. Type exit to exit.")

while user_question != "exit":
  compiled_graph.invoke({ "messages": [ HumanMessage(content=user_question) ] })
  user_question=input("How can i help you. Type exit to exit.")




```

```python
#!pip install langgraph
from langgraph.graph import START , END , StateGraph , add_messages
from typing import Union, TypedDict, Sequence,Annotated
import os
#!pip install langchain_openai
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage,AIMessage,BaseMessage
from google.colab import userdata
os.environ['OPENAI_API_KEY']=userdata.get('ai_agents_openai')
llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0)

class MemoryAgent(TypedDict):
  #messages: list[Union[HumanMessage,AIMessage]]
  messages:Annotated[Sequence[BaseMessage], add_messages]

def model_invoke(state: MemoryAgent)-> MemoryAgent:
  response=llm.invoke(state["messages"])
  ## This is stateful call . Not a best practice
  #state["messages"].append(response.content)
  ## This is a stateless call . Best practice
  print(response.content)
  #return {"messages": state["messages"] + [AIMessage(content=response.content)]}
  ## If we want to use add_messages





graph=StateGraph(MemoryAgent)
graph.add_node("action",model_invoke)
graph.add_edge(START,"action")
graph.add_edge("action",END)

compiled_graph=graph.compile()

chat_conversation=[]

user_question=input("How can i help you. Type exit to exit.")

while user_question != "exit":
  chat_conversation.append(HumanMessage(content=user_question))
  compiled_graph.invoke({ "messages": chat_conversation })
  user_question=input("How can i help you again. Type exit to exit.")



```
